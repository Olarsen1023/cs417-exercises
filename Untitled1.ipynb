```python
%%writefile code.py

# tests if a number is less than or greater
# based on input from comand line
import sys
new_lst = [ x for x in sys.argv[1:]]
if new_lst[1] == "Less Than":
  if new_lst[0] < new_lst[2]:
    print("True")
elif new_lst[1] == "Greater Than":
  if new_lst[0] > new_lst[2]:
    print("True")
elif new_lst[1] == "Equal To":
  if new_lst[0] == new_lst[2]:
    print("True")
else:
  print("False")
```
