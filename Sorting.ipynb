```python
import random

# Create an unsorted list with 1000 random elements
unsorted_list = [random.randint(0, 1000) for _ in range(1000)]

# Create a sorted list by sorting a copy of the unsorted list
sorted_list = sorted(unsorted_list)
```

```python
def binary_search(arr, target):
  first = 0
  last = len(arr) - 1

  while first <= last:
    mid = (first + last) // 2
    if arr[mid] == target:
      return True
    elif arr[mid] < target:
      last = mid - 1
      print(last)
    else:
      first = mid + 1
      print(first)
  return False
```

```python
binary_search(sorted_list, 500)
binary_search(unsorted_list, 200)
```
