```python
class fraction():
  def __init__(self, top, bottom):
    self.num = top
    self.denom = bottom

  def __str__(self):
    return f"{self.num} / {self.denom}"

f1 = fraction(3,4)
f2 = fraction(7,8)
```

```python
print(f1)
print(f2)
```
